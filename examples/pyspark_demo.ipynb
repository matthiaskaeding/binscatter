# Large PySpark Binscatter Demo

This example creates ten million rows directly in Spark and passes the lazy Spark dataframe to `binscatter`. Spark performs the filtering and aggregation; only the binned result is collected for plotting.

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

from binscatter import binscatter

spark = (
    SparkSession.builder.master("local[*]")
    .appName("binscatter-pyspark-demo")
    .getOrCreate()
)
```

```python
# Ten million rows are generated and transformed entirely in Spark.
spark_data = (
    spark.range(10_000_000)
    .withColumn("x", F.col("id").cast("double") / 100_000)
    .withColumn("y", F.sin(F.col("x")) + F.col("x") / 20)
)

binscatter(
    spark_data, x="x", y="y", num_bins=100, title="Ten million PySpark rows"
).show()
```
